## Section 4 — Imports


```python
import matplotlib.pyplot as plt

from sklearn.metrics import (
    ConfusionMatrixDisplay,
    RocCurveDisplay,
    PrecisionRecallDisplay
)
```

## Section 5 — Confusion matrix


```python
def plot_confusion_matrix(y_true, y_prob, threshold=0.5):
    y_pred = (y_prob >= threshold).astype(int)

    ConfusionMatrixDisplay.from_predictions(y_true, y_pred, display_labels=["Stage 1-3", "stage 4"])
    # ConfusionMatrixDisplay.from_predictions(...)

    plt.title("Confusion Matrix")
    plt.show()
```

## Section 6 — ROC curve


```python
def plot_roc_curve(y_true, y_prob):
    RocCurveDisplay.from_predictions (y_true, y_prob)
    # RocCurveDisplay.from_predictions(...)

    plt.title("ROC Curve")
    plt.show()
```

## Section 7 — Precision-recall curve


```python
def plot_pr_curve(y_true, y_prob):
    PrecisionRecallDisplay.from_predictions (y_true, y_prob)

    plt.title("Precision-Recall Curve")
    plt.show()
```

## Section 8 — Combined plotting helper


```python
def plot_all_evaluation_curves(
    y_true,
    y_prob,
    model_name="Model"
):
    print(f"Evaluation plots: {model_name}")

    plot_confusion_matrix(y_true, y_prob)
    plot_roc_curve(y_true, y_prob)
    plot_pr_curve(y_true, y_prob)
```
